### Parameters

This notebook describes how to use the parameters class in `jasmine_toolkit`.

The `parameters` module can be imported as below.

```python
import jasmine_toolkit.parameters as p
p?
```

The module has several submodules. `parameters.detector` contains information related to detectors.

```python
p.detector?
```

You can retrieve pre-defined values. For example, a physical pixel size is defined as `parameters.detector.pizel_size`.

```python
p.detector.pixel_size
```

Detailed information can be shown like this.

```python
p.detector.pixel_size?
```

Some parameters are calculated from others (see reference).

```python
p.detector.naxis1?
```

`parameters.telescope.pupil_diameter` defines the diameter of the entrance pupil.

```python
p.telescope.pupil_diameter?
```

`parameters.telescope.effective_focal_length` provides the effective focal length of the telescope, which depends on `pupil_diameter` and `f_ratio`. Thus, `effective_focal_length` is defined as a calculated parameter (function).

```python
p.telescope.effective_focal_length
```

Parameter values should be carefully modified. `parameters.update_parameters` provides a context where a new value can be assigned to a parameter without warnings. In the example below, a new pupil diameter is assigned.

```python
import astropy.units as u

with p.update_parameters():
  p.telescope.pupil_diameter = 25 * u.cm

p.telescope.pupil_diameter?
```

Since `effective_focal_length` depends on `pupil_diameter`, `effective_focal_length` returns an updated value.

```python
p.telescope.effective_focal_length?
```
